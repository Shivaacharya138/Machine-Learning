**PCA**

```python
# 1. STANDARDIZATION


from sklearn.preprocessing import StandardScaler

```

```python
x_std = StandardScaler().fit_transform(x)
```

```python
x_std
```

```python
x1 = x_std.T

x1   #
```

```python
x1.shape
```

```python
# covariance :


cov_mat = np.cov(x1)
```

```python
cov_mat
```

```python
# eign vector : direction through we get the max variance

# eigen value :  how much variance is  being captured
eign_val ,eign_vec = np.linalg.eig(cov_mat)
```

```python
eign_val


#
```

```python
eign_vec
```

```python
total = sum(eign_val)
```

```python
var_exp = []
for i in eign_val:
  percentage = (i/total)*100

  var_exp.append(percentage)
```

```python
var_exp

# pc1 + pc2 = 21%
# pc1 + pc2+ pc3+pc4 =


#  95%
```

```python
cum_sum_var = np.cumsum(var_exp)

```

```python
cum_sum_var


# i need to take those components which are giving maximum infor about my data
```

```python
plt.figure(figsize=(10,5))
plt.bar(range(len(var_exp)), var_exp, label='indivdual explaned variance', color='g')
plt.step(range(len(cum_sum_var)), cum_sum_var, label ='cumulative explained variance')
plt.ylabel('explained variance')
plt.xlabel('principal component')
plt.legend()
plt.show()
```

```python
# select the prinicpal components from 20 to 30 : they are able to capture above 80 % of the
# information

# afer 85 , there is no any information left to capture  that's why the line is straight
```

```python
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
```

```python
pca =PCA(n_components= 40)
```

```python
pca_x = pca.fit_transform(x_std)
```

```python
pca_x.shape

```

```python
# model implementation

x_train, x_test , y_train , y_test  = train_test_split(pca_x, y, train_size=0.8, random_state = 23)
```

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(random_state = 23)

dt.fit(x_train, y_train)
```

```python
predict = dt.predict(x_test)
```

```python
predict
```

```python
df1 = pd.DataFrame({'Actual': y_test[0], 'Predicted': predict})
```

```python
df1
```

```python
from sklearn.metrics import *

accuracy_score(y_test , predict)
```

```python
# logistic

from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()

lr.fit(x_train, y_train)

y_pred_lr = lr.predict(x_test)


lr_acc = accuracy_score(y_test, y_pred_lr)

print("logistics regression accuracy", lr_acc)
```

```python
# random forest

from sklearn.ensemble import RandomForestClassifier

rf  = RandomForestClassifier()
rf.fit(x_train, y_train)

rf_predict = rf.predict(x_test)

accuracy_score(y_test, rf_predict)
```
